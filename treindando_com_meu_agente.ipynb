# Primeira lição percorrer lista usando o for.

```python

linguagens = ["Python", "JavaScript", "C++", "Java"]

for linguagem in linguagens:
    print(f"Eu estou a aprender {linguagem}!")
```

Explicação Detalhada do Código
linguagens = [...]: Aqui definimos a nossa lista. Os elementos estão guardados entre parênteses retos [] e separados por vírgulas.

for linguagem in linguagens:: Esta é a linha mágica. Lemos isto como: "Para cada linguagem dentro da lista linguagens, faz o seguinte...".

linguagens é a lista que já existia.

linguagem (no singular) é o nome que escolhemos para o item atual. Podes mudar este nome para o que quiseres, como item ou elemento.

O : (dois pontos) no final é obrigatório. Diz ao Python que o bloco de código que vem a seguir pertence ao ciclo.

O Espaço (Indentação): Reparaste que a linha do print tem uns espaços em branco antes de começar? Isso chama-se indentação. No Python, esses espaços servem para dizer ao programa que aquela linha só deve ser executada dentro do ciclo for.
